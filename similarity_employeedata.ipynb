```python
#creating a new collection

books_collection = client.get_or_create_collection(
    name = "books_collection_name",
    configuration={
        "hnsw":{"space":"cosine"},
        "embedding_functions":ef
    }
)
print("books collection created")
```

```python
#Add data to the collection
books_documents = []
for book in books:
    book_document = "book title: "+book['title'] + " written by "+ book['author'] + " , published on " + str(book['year']) + " , which talks about " + book['description']
    books_documents.append(book_document)

books_collection.add(
    ids = [book['id'] for book in books],
    documents = books_documents,
    metadatas = [{
        "title": book['title'],
        "author": book['author'],
        "genre": book['genre'],
        "year": book['year'],
        "rating": book['rating'],
        "pages": book['pages'],
        "themes": book['themes'],
        "setting": book['setting']
        }for book in books
    ]
)
```

```python
query_text = "magical fantasy adventure"

results = books_collection.query(
    query_texts = [query_text],
    where = {"genre" : "Fantasy or Science Fiction"},
    n_results = 3
)
print(f"Query: '{query_text}'")
for i, (doc_id, document, distance) in enumerate(zip(
    results['ids'][0], results['documents'][0], results['distances'][0]
)):
    metadata = results['metadatas'][0][i]
    print(f"  {i+1}. {metadata['title']} ({doc_id}) - Distance: {distance:.4f}")
    print(f"     year: {metadata['year']}, author: {metadata['author']}")
    print(f"     Document: {document[:100]}...")
```
